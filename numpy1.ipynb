# Row slicing 1:3
* row 0  →  skipped   (before 1)
* row 1  →  ✅ included
* row 2  →  ✅ included
* row 3  →  skipped   (3 is excluded)

```python
print(array_3[1:3, 1:])
```

# Column slicing
* col 0  →  skipped   (before 1)
* col 1  →  ✅ included
* col 2  →  ✅ included (no end limit means go to last)

```python
array_3.sum()
```
