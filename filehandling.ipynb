```python
import os

# Create a dummy file for demonstration
file_name = "old_file.txt"
with open(file_name, "w") as f:
  f.write("This is a test file.")
print(f"Created file: {file_name}")

# Rename the file
new_file_name = "new_file.txt"
os.rename(file_name, new_file_name)
print(f"Renamed '{file_name}' to '{new_file_name}'")

# Verify if the new file exists
if os.path.exists(new_file_name):
  print(f"'{new_file_name}' exists.")

# Read content from the renamed file
with open(new_file_name, 'r') as f:
    content = f.read()
print(f"Content of '{new_file_name}': {content}")

# Remove the file
os.remove(new_file_name)
print(f"Removed file: {new_file_name}")

# Verify if the file is removed
if not os.path.exists(new_file_name):
  print(f"'{new_file_name}' no longer exists.")
```
